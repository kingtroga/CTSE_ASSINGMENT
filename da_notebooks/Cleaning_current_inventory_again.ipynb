```python
# Check the data types and basic info
print("Data types:")
print(df.dtypes)
print("\n" + "="*50 + "\n")

print("DataFrame info:")
print(df.info())
print("\n" + "="*50 + "\n")

print("First few rows to see the structure:")
print(df.head(3))
```

### The main problems
1. **Column names are wrong** - they're showing the summary numbers instead of proper names
2. **The actual column names are in the first row** - (Product Name, msku, Opening Stock, etc.)
3. **All data is stored as 'object' type** - instead of proper numeric types.
4. **The first row contains headers, not data**



## Step 2: Fixing the Column Names and removing the header row


```python
# First, let's see what the actual column names should be (they're in row 0)
print("Current first row (which contains our real column names):")
print(df.iloc[0].tolist())
print("\n" + "="*50 + "\n")

# Set the first row as column names and remove it from the data
new_column_names = df.iloc[0].tolist()
print("New column names will be:")
print(new_column_names)
print("\n" + "="*50 + "\n")

# Create a new dataframe with proper column names
df_cleaned = df.iloc[1:].copy()  # Take all rows except the first one
df_cleaned.columns = new_column_names  # Set the proper column names

# Reset the index so it starts from 0 again
df_cleaned = df_cleaned.reset_index(drop=True)

print("After fixing column names:")
print("Shape:", df_cleaned.shape)
print("\nFirst few rows:")
print(df_cleaned.head(3))
```

## Why are there still 1011 rows?

I know one row is the head row but did we still include the summary row in the data?

If there are supposed to be 1009 products but we have 1011 rows after removing the header, that means there are likely 2 extra rows that aren't actual product data.

Let's investigate. 


### Step 2.5:  Check for summary or non-product rows

```python
# Check for any rows where Product Name might be a number (summary rows)
print("Checking for rows where Product Name contains only numbers:")
print("\n" + "="*50 + "\n")

# Check if any Product Name values look like numbers
numeric_product_names = []
for idx, product_name in enumerate(df_cleaned['Product Name']):
    # Convert to string and strip whitespace, then check if it's numeric
    str_name = str(product_name).strip()
    if str_name.isdigit() or str_name.replace('.', '').isdigit():
        numeric_product_names.append((idx, product_name))

print("Rows with numeric Product Names:")
if numeric_product_names:
    for idx, name in numeric_product_names:
        print(f"Row {idx}: Product Name = '{name}'")
        print(df_cleaned.iloc[idx])
        print("-" * 30)
else:
    print("No numeric Product Names found")

print("\n" + "="*50 + "\n")

# Also check the very end for any potential summary rows
print("Let's examine the last few rows more carefully:")
for i in range(max(0, len(df_cleaned)-5), len(df_cleaned)):
    row = df_cleaned.iloc[i]
    print(f"Row {i}: '{row['Product Name']}' | MSKU: '{row['msku']}' | Opening Stock: '{row['Opening Stock']}'")
```

### Findings

1. **No numeric Product Names found** - So no summary rows to remove
2. **All rows at the end look legitimate** - proper product names and MSKUs 
3. **The "1009" in the original summary was incorrect** - CSTE actually has **1011 products** in it's inventory not 1009

**1011 rows are valid product data, and we don't need to remove any rows. The discrepancy was in the original summary information.**

## Step 3: Fix the data types 

All columns are stored as 'object' (strings), we need to convert the numeric columns to proper integer types

```python
# First, let's verify all MSKUs are unique (they should be)
print("Verifying MSKU uniqueness:")
print(f"Total rows: {len(df_cleaned)}")
print(f"Unique MSKUs: {df_cleaned['msku'].nunique()}")
print("All MSKUs are unique:", len(df_cleaned) == df_cleaned['msku'].nunique())
print("\n" + "="*50 + "\n")

# Replace NaN in Buffer Stock with 0 as mentioned
df_cleaned['Buffer Stock'] = df_cleaned['Buffer Stock'].fillna(0)

# Convert numeric columns to integers
numeric_columns = ['Opening Stock', 'Buffer Stock'] + [col for col in df_cleaned.columns if col not in ['Product Name', 'msku']]

print("Converting these columns to integers:")
print(numeric_columns)
print("\n" + "="*30 + "\n")

# Convert to integers
for col in numeric_columns:
    df_cleaned[col] = pd.to_numeric(df_cleaned[col], errors='coerce').astype(int)

print("After conversion - Data types:")
print(df_cleaned.dtypes)
print("\nFirst few rows with proper data types:")
print(df_cleaned.head(3))
```

# The Data Cleaning is now **Complete**

## Step 4: Data Overview

```python
# Let's get a final overview of our clean data
print("=== FINAL CLEANED DATASET OVERVIEW ===")
print(f"Shape: {df_cleaned.shape}")
print(f"Total Products: {len(df_cleaned)}")
print("\nColumn Summary:")
print("Text columns:", ['Product Name', 'msku'])
warehouse_cols = [col for col in df_cleaned.columns if col not in ['Product Name', 'msku', 'Opening Stock', 'Buffer Stock']]
print("Warehouse columns:", warehouse_cols)

print("\n=== STOCK SUMMARY ===")
print("Opening Stock:")
print(f"  Total Opening Stock: {df_cleaned['Opening Stock'].sum():,}")
print(f"  Products with stock > 0: {(df_cleaned['Opening Stock'] > 0).sum()}")
print(f"  Products with zero stock: {(df_cleaned['Opening Stock'] == 0).sum()}")

print("\nBuffer Stock:")
print(f"  Total Buffer Stock: {df_cleaned['Buffer Stock'].sum():,}")

print("\nWarehouse Stock (TLCQ to XHJ9):")
warehouse_total = df_cleaned[warehouse_cols].sum().sum()
print(f"  Total across all warehouses: {warehouse_total:,}")
print(f"  Individual warehouse totals:")
for col in warehouse_cols:
    warehouse_sum = df_cleaned[col].sum()
    print(f"    {col}: {warehouse_sum:,}")

# Check for negative values (backorders)
print("\n=== NEGATIVE STOCK CHECK ===")
negative_opening = (df_cleaned['Opening Stock'] < 0).sum()
negative_warehouses = sum((df_cleaned[col] < 0).sum() for col in warehouse_cols)
print(f"Products with negative Opening Stock: {negative_opening}")
print(f"Warehouse locations with negative stock: {negative_warehouses}")

print("\nFirst 3 rows of clean data:")
print(df_cleaned.head(3))

# Save the cleaned DataFrame to a new CSV file
output_dir = notebook_dir.parent / "clean_data"
output_dir.mkdir(parents=True, exist_ok=True)
output_file = output_dir / "cleaned_inventory.csv"

df_cleaned.to_csv(output_file, index=False)
```

Buffer Stock is typically "safety stock" or "reserve stock" - extra inventory kept on hand to prevent stockouts due to demand fluctuations or supply delays. CSTE's total of 350 units suggests a conservative buffer strategy.
