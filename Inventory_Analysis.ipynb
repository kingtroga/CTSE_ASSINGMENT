# Inventory Analysis

The purpose of this notebook is to understand the inventory data and also clean and prep the data.


```python
import pandas as pd
import numpy as np

# Load the data
print("🔍 Loading current_inventory.csv...")
df = pd.read_csv('current_inventory.csv')

# Basic info
print(f"✅ Data loaded successfully!")
print(f"📊 Shape: {df.shape} (rows, columns)")
print(f"📋 Columns: {list(df.columns)}")

# Show first few rows
print("\n📋 First 3 rows:")
print(df.head(3))
```

```python
# Fix the header issue - use the first row as column names
print("🔧 Fixing column headers...")

# Get the actual column names from the first row
actual_columns = df.iloc[0].tolist()
print(f"📋 Actual columns found: {actual_columns}")

# Set these as the column names and drop the header row
df.columns = actual_columns
df = df.drop(0).reset_index(drop=True)

print(f"\n✅ Headers fixed!")
print(f"📊 New shape: {df.shape}")
print(f"📋 Corrected columns: {list(df.columns)}")

# Show first few rows with correct headers
print("\n📋 First 3 rows with correct headers:")
print(df.head(3))
```

### Discovery

I count 15 warehouse locations across Indian cities:

1. BLR = Bangalore (BLR7, BLR8)
2. BOM = Mumbai/Bombay (BOM5, BOM7)
3. CCU/CCX = Kolkata/Calcutta (CCU1, CCX1)
4. DEL/DEX = Delhi (DEL4, DEL5, DEX3)
5. PNQ = Pune (PNQ2, PNQ3)
6. TLCQ, SDED, SDEE, XHJ9 = Other cities

Now let's analyze the data types and stock distribution:

```python
# Check data types and convert warehouse columns to numeric
print("🔍 Checking data types...")
print(df.dtypes)

print("\n🔢 Converting warehouse columns to numeric...")

# List all warehouse columns (everything except Product Name, msku, Opening Stock, Buffer Stock)
warehouse_cols = ['TLCQ', 'BLR7', 'BLR8', 'BOM5', 'BOM7', 'CCU1', 'CCX1', 
                  'DEL4', 'DEL5', 'DEX3', 'PNQ2', 'PNQ3', 'SDED', 'SDEE', 'XHJ9']

# Convert to numeric, replacing any non-numeric values with 0
for col in warehouse_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

print("✅ Conversion complete!")

# Also convert Opening Stock to numeric
df['Opening Stock'] = pd.to_numeric(df['Opening Stock'], errors='coerce').fillna(0)

print("\n📊 Basic statistics for key columns:")
print("="*50)
print("Opening Stock:")
print(df['Opening Stock'].describe())

print(f"\n📦 Sample of warehouse stock data:")
print(df[['Product Name'] + warehouse_cols[:5]].head(3))
```

### 📊 Quick Insights:

- 1,011 products in inventory
- Average stock: 6.6 units per product
- 75% of products have ≤2 units (very lean inventory!)
- Max stock: 624 units (one popular product!)
- ⚠️ Negative stock spotted: BOM7 shows -1 for Chimmy Pillow

Now let's analyze the warehouse stock distribution and find those problem areas:

```python
# Warehouse stock analysis
warehouse_cols = ['TLCQ', 'BLR7', 'BLR8', 'BOM5', 'BOM7', 'CCU1', 'CCX1', 
                  'DEL4', 'DEL5', 'DEX3', 'PNQ2', 'PNQ3', 'SDED', 'SDEE', 'XHJ9']

print("🏭 WAREHOUSE STOCK ANALYSIS")
print("="*50)

# Total stock per warehouse
print("📦 Total stock per warehouse:")
warehouse_totals = df[warehouse_cols].sum().sort_values(ascending=False)
for warehouse, total in warehouse_totals.items():
    print(f"  {warehouse}: {total:>8.0f} units")

print(f"\n🎯 TOTAL INVENTORY ACROSS ALL WAREHOUSES: {warehouse_totals.sum():,.0f} units")

# Calculate total stock per product
df['total_stock'] = df[warehouse_cols].sum(axis=1)

print(f"\n📊 STOCK DISTRIBUTION:")
print(f"  Products with stock > 0: {len(df[df['total_stock'] > 0]):>4}")
print(f"  Products with zero stock: {len(df[df['total_stock'] == 0]):>4}")  
print(f"  Products with negative stock: {len(df[df['total_stock'] < 0]):>4}")

# Find negative stock issues
print(f"\n⚠️  NEGATIVE STOCK ALERTS:")
print("="*35)
for col in warehouse_cols:
    negative_items = df[df[col] < 0]
    if len(negative_items) > 0:
        print(f"\n🚨 {col}: {len(negative_items)} items with negative stock")
        for idx, row in negative_items.iterrows():
            print(f"    {row['Product Name']}: {row[col]}")

# Top 5 products by total stock
print(f"\n🏆 TOP 5 PRODUCTS BY TOTAL STOCK:")
print("="*40)
top_stock = df.nlargest(5, 'total_stock')[['Product Name', 'msku', 'total_stock']]
for idx, row in top_stock.iterrows():
    print(f"  {row['Product Name']}: {row['total_stock']:.0f} units")

# Products with zero stock everywhere
zero_stock = df[df['total_stock'] == 0]
print(f"\n📋 ZERO STOCK ANALYSIS:")
print(f"  {len(zero_stock)} products have no stock anywhere ({len(zero_stock)/len(df)*100:.1f}%)")
print(f"  Sample zero-stock products:")
for i, row in zero_stock.head(5).iterrows():
    print(f"    {row['Product Name']}")
```

## 🏭 MAJOR FINDINGS:
### 📊 Inventory Concentration Crisis:

- 99.5% of stock is in TLCQ warehouse only (5,755 out of 5,726 units)
- All other warehouses are essentially empty or have negative stock
- Single point of failure - entire business depends on TLCQ

### ⚠️ SERIOUS Negative Stock Problem:

- 27 products oversold across 6 warehouses
- Worst case: HP Brown MB is -22 units (sold 22 more than available!)
- Data sync issues between sales and inventory systems

### 📋 Stock Portfolio Health:

- 64.7% of products (654/1011) have ZERO stock everywhere
- Only 32.6% (330/1011) products actually available to sell
- Business continuity risk - most catalog unavailable

### 🏆 Star Products:

- ThugLife Thin: 603 units (10.5% of total inventory!)
- HBD Brown MB: 380 units
- Harry Potter Round SG: 347 units

```python
# MSKU Pattern Analysis
print("🏷️  MSKU PATTERN ANALYSIS")
print("="*50)

# Split MSKU into components
print("📋 Analyzing MSKU structure...")
df['msku_clean'] = df['msku'].astype(str)

# Split by underscore
msku_parts = df['msku_clean'].str.split('_', expand=True)
print(f"MSKU components found: {msku_parts.shape[1]} parts")

# Analyze each component
if msku_parts.shape[1] >= 4:
    df['prefix'] = msku_parts[0]  # Should be CSTE
    df['number'] = msku_parts[1]  # Sequential number
    df['category'] = msku_parts[2]  # Product category (MB, ST, etc.)
    df['description'] = msku_parts[3]  # Product description
    
    print(f"\n🔍 MSKU Component Analysis:")
    print(f"="*35)
    
    print(f"Prefix analysis:")
    print(df['prefix'].value_counts().head())
    
    print(f"\nCategory codes (what do they mean?):")
    category_counts = df['category'].value_counts()
    for cat, count in category_counts.items():
        print(f"  {cat}: {count:>3} products")
    
    print(f"\n📊 Category Distribution with Stock Status:")
    print("="*45)
    for cat in category_counts.index:
        cat_data = df[df['category'] == cat]
        has_stock = len(cat_data[cat_data['total_stock'] > 0])
        zero_stock = len(cat_data[cat_data['total_stock'] == 0])
        negative_stock = len(cat_data[cat_data['total_stock'] < 0])
        avg_stock = cat_data['total_stock'].mean()
        
        print(f"{cat:>4}: {len(cat_data):>3} total | "
              f"{has_stock:>3} stocked | {zero_stock:>3} zero | "
              f"{negative_stock:>2} negative | avg: {avg_stock:>5.1f}")

# Look at number patterns
print(f"\n🔢 MSKU Number Range:")
if 'number' in df.columns:
    # Convert to numeric for analysis
    df['number_numeric'] = pd.to_numeric(df['number'], errors='coerce')
    valid_numbers = df['number_numeric'].dropna()
    print(f"  Lowest: {valid_numbers.min():.0f}")
    print(f"  Highest: {valid_numbers.max():.0f}")
    print(f"  Range: {valid_numbers.max() - valid_numbers.min():.0f}")
    print(f"  Valid numbers: {len(valid_numbers)}/{len(df)}")

# Sample MSKUs by category
print(f"\n📋 Sample MSKUs by Category:")
print("="*35)
for cat in category_counts.head(5).index:
    sample = df[df['category'] == cat]['msku'].iloc[0]
    print(f"  {cat}: {sample}")
```
